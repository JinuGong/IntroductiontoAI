## 1. 함수 정의

```python
def 함수이름(매개변수):
    실행할_코드
```

````python
def print_address():
    print('서울특별시 성북구 삼선교로 16길 116')
    print('한성대학교')

print_address()
print_address()    # 여러 번 재사용 가능
````

## 2. 매개변수와 인자

- **매개변수 (parameter)** : 함수 정의의 `def f(name):` 에서 `name`
- **인자 (argument)** : 호출할 때 넘기는 값 `f('Alice')` 의 `'Alice'`

````python
def greet(name):              # name 은 매개변수
    print(f'안녕하세요, {name}님!')

greet('Alice')                # 'Alice' 는 인자
greet('Bob')
````

---
## 3. `return` — 값 돌려주기

`return` 으로 결과 값을 반환한다. `return` 이 없으면 `None` 을 반환.

````python
def circle_area(radius):
    return 3.14 * radius ** 2

r = 5
area = circle_area(r)
print(area)

# return 없는 함수의 반환값
def nothing():
    print('print만 하고 return 없음')
result = nothing()
print(result)        # None
````

### 3.1 매개변수가 여러 개일 때

````python
def get_sum(start, end):
    """start ~ end 까지의 합을 반환."""
    return sum(range(start, end + 1))

print(get_sum(1, 100))
````

---
## 4. 여러 값 반환

파이썬에서는 **튜플**로 묶어 한 번에 여러 값을 반환할 수 있다.

````python
def sort_two(a, b):
    """두 값을 (작은값, 큰값) 으로 반환."""
    if a > b:
        return b, a
    return a, b

low, high = sort_two(7, 3)
print(low, high)
````

### 4.1 연습: 최대공약수와 최소공배수

유클리드 호제법 사용.

````python
def gcd_lcm(a, b):
    x, y = a, b
    while y:
        x, y = y, x % y      # 유클리드 호제법
    gcd = x
    lcm = a * b // gcd
    return gcd, lcm

g, l = gcd_lcm(12, 18)
print(f'GCD={g}, LCM={l}')
````

## 5. 변수의 스코프 (scope)

함수 안에서 만든 변수는 **함수 안에서만** 살아있다. 함수 밖 변수를 함수 안에서 **수정** 하려면 `global` 키워드가 필요하다.

````python
counter = 0

def increase():
    counter = 1          # 새 지역 변수 — 바깥 counter 는 그대로
    print('함수 안:', counter)

increase()
print('함수 밖:', counter)
````

````python
counter = 0

def increase():
    global counter       # 바깥 counter 를 가리키도록
    counter += 1
    print('함수 안:', counter)

increase()
increase()
print('함수 밖:', counter)
````

> ⚠️ `global` 은 가능한 한 피하자. 상태가 어디서 바뀌는지 추적이 어려워진다. 보통은 **값을 반환**하고 호출자가 다시 대입하는 게 깔끔하다.

## 6. 디폴트 인자

매개변수에 기본값을 줄 수 있다. 호출 시 값을 안 넘기면 기본값이 쓰인다.

````python
def order(num, pickle=True, onion=True):
    print(f'햄버거 {num}개 (피클: {pickle}, 양파: {onion})')

order(1)                              # 기본값 사용
order(2, pickle=False)                # 일부만 변경
order(3, pickle=False, onion=False)   # 모두 지정
````

> ⚠️ 디폴트 인자는 매개변수 목록 **뒤쪽**에 와야 한다.
> `def f(a=1, b)` ❌ → `def f(b, a=1)` ✅

## 7. 위치 인자와 키워드 인자

- **위치 인자** : 정의된 순서대로 값을 매핑
- **키워드 인자** : `name=value` 로 명시적으로 매핑

````python
def power(base, exponent):
    return base ** exponent

print(power(2, 10))                # 위치: base=2, exponent=10
print(power(base=2, exponent=10))  # 키워드
print(power(exponent=10, base=2))  # 키워드면 순서 바꿔도 OK

# 섞어 쓸 수 있지만, 키워드는 위치 다음에
print(power(2, exponent=10))       # ✅
# print(power(base=2, 10))         # ❌ SyntaxError
````

## 8. 재귀 함수 (recursion)

함수가 **자기 자신을 호출**한다. 반드시 **종료 조건(base case)** 이 있어야 무한 재귀를 피할 수 있다.

````python
def factorial(n):
    if n <= 1:           # 종료 조건
        return 1
    return n * factorial(n - 1)

print(factorial(5))      # 120
print(factorial(10))
````

### 8.1 연습: 피보나치 수열

$f(0)=0, f(1)=1, f(n) = f(n-1) + f(n-2)$

````python
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

for i in range(10):
    print(fib(i), end=' ')
````

> ⚠️ 단순 재귀 피보나치는 매우 느리다 (지수 시간). 큰 n 은 **메모이제이션** 이나 반복문으로 풀자.

## 9. `lambda` — 한 줄 익명 함수

간단한 함수를 변수에 담거나 다른 함수에 넘길 때 편하다.

```python
lambda 매개변수: 표현식
```

````python
add = lambda x, y: x + y
print(add(2, 3))

# sorted 의 key 매개변수에 자주 쓴다
words = ['banana', 'apple', 'fig', 'cherry']
print(sorted(words, key=lambda w: len(w)))   # 길이순 정렬

# 딕셔너리 값 기준 정렬
scores = {'A': 85, 'B': 92, 'C': 78}
top = sorted(scores.items(), key=lambda kv: -kv[1])
print(top)
````
