## 3. 변수와 자료형

변수는 값을 담는 그릇이다. 파이썬은 **자료형을 자동으로 추론**한다.

| 자료형 | 예시 | 설명 |
|--------|------|------|
| `int`   | `3`, `-7` | 정수 |
| `float` | `3.14`, `1.0` | 실수 |
| `str`   | `'hi'`, `"hello"` | 문자열 |
| `bool`  | `True`, `False` | 참/거짓 |
| `list`  | `[1, 2, 3]` | 리스트 (06장) |

````python
x = 3
name = 'Jinu'
accuracy = 0.99
is_student = True

print(type(x))
print(type(name))
print(type(accuracy))
print(type(is_student))
````

> ⚠️ `True`, `False` 는 **첫 글자 대문자**다. `true` 는 에러.

### 3.1 자료형 확인: `type()`

````python
print(type(2 * 3))       # int
print(type(2.0 * 3))     # float — 하나라도 실수면 결과는 실수
print(type([2, 2]))      # list
````

---
## 4. 형 변환

문자열과 숫자는 바로 더할 수 없다. 명시적으로 변환해야 한다.

````python
print('100' + '200')   # 문자열 + 문자열 → 이어붙임: '100200'
# print('100' + 100)   # ❌ TypeError: 문자열과 숫자를 합칠 수 없다
print('100' + str(100))  # ✅ 정수를 문자열로 변환
````

### 4.1 변환 함수

| 함수 | 의미 |
|------|------|
| `int(x)` | 정수로 변환 |
| `float(x)` | 실수로 변환 |
| `str(x)` | 문자열로 변환 |
| `bool(x)` | 불리언으로 변환 |

````python
print(int(3.7))      # 3 (소수점 버림)
print(int('42'))     # 42
print(float('3.14')) # 3.14
print(str(123))      # '123'
print(int(float('6.0')))  # 6
````

> ⚠️ `int('6.3')` 은 에러. 정수로 가려면 먼저 `float()` 을 거쳐야 한다.

---
## 5. 문자열 다루기

파이썬은 작은따옴표 `'...'` 와 큰따옴표 `"..."` 둘 다 문자열로 인식한다. 따옴표 종류를 섞으면 문자열 안에 따옴표를 넣을 수 있다.

````python
sentence1 = "This is 'python for AI' class"
sentence2 = 'This is "python for AI" class'
sentence3 = """여러 줄 문자열도
이렇게 작성할 수 있다."""

print(sentence1)
print(sentence2)
print(sentence3)
````

### 5.1 이스케이프 시퀀스

같은 종류의 따옴표를 안에 넣으려면 `\` (백슬래시) 로 탈출한다.

| 시퀀스 | 의미 |
|--------|------|
| `\\\"` | 큰따옴표 |
| `\\\'` | 작은따옴표 |
| `\\n` | 줄바꿈 |
| `\\t` | 탭 |

````python
sentence = "This is \"python for AI\" class"
print(sentence)

multiline = "첫 줄\n둘째 줄\t(탭)"
print(multiline)
````

### 5.2 문자열 곱하기 / 더하기

````python
print('안녕 ' * 3)        # '안녕 안녕 안녕 '
print('AI' + ' 기초')      # 'AI 기초'
````
