## 1. 산술 연산자

| 연산자 | 의미 | 예시 |
|--------|------|------|
| `+` `-` `*` `/` | 사칙연산 | `5 / 2 → 2.5` |
| `//` | 몫 (정수 나눗셈) | `5 // 2 → 2` |
| `%` | 나머지 | `5 % 2 → 1` |
| `**` | 거듭제곱 | `2 ** 10 → 1024` |

```python
print(5 + 2, 5 - 2, 5 * 2, 5 / 2)
print(5 // 2, 5 % 2)
print(2 ** 10)
print(2 ** 0.5)   # 제곱근 (= √2)
```

### 1.1 몫과 나머지 활용

초를 시·분·초로 변환하기처럼, 단위 변환에 자주 쓴다.

```python
# 3725초 = 1시간 2분 5초
seconds = 3725
h = seconds // 3600
m = (seconds % 3600) // 60
s = seconds % 60
print(f'{h}시간 {m}분 {s}초')
```

## 2. 복합 할당 연산자

`a = a + 1` 처럼 자기 자신을 갱신할 때, `a += 1` 로 줄여 쓸 수 있다.

| 표현 | 같은 의미 |
|------|----------|
| `a += 5` | `a = a + 5` |
| `a -= 5` | `a = a - 5` |
| `a *= 2` | `a = a * 2` |
| `a //= 2` | `a = a // 2` |

```python
a = 1000
a += 100   # 1100
a *= 2     # 2200
a //= 3    # 733
print(a)
```

## 3. 비교 연산자

비교 결과는 항상 `bool` (`True` / `False`).

| 연산자 | 의미 |
|--------|------|
| `==` | 같다 |
| `!=` | 다르다 |
| `<`, `>` | 작다, 크다 |
| `<=`, `>=` | 작거나 같다, 크거나 같다 |

```python
a = 100
print(a == 100)     # True
print(a != 100)     # False
print(0 < a <= 100) # True — 파이썬은 연쇄 비교를 지원
```

## 4. 논리 연산자

`and`, `or`, `not` 을 써서 `bool` 값을 조합한다.

| 연산자 | 의미 |
|--------|------|
| `a and b` | 둘 다 참 → 참 |
| `a or b` | 하나라도 참 → 참 |
| `not a` | 참 ↔ 거짓 |

```python
print(True and False)   # False
print(True or False)    # True
print(not True)         # False

# 숫자도 bool로 평가됨: 0과 빈 문자열은 False
print(bool(0), bool(1), bool(''), bool('a'))
```

### 4.1 단락 평가 (short-circuit evaluation)

`and` 는 앞이 거짓이면 뒤를 보지 않고, `or` 는 앞이 참이면 뒤를 보지 않는다.
파이썬에서는 **마지막으로 평가된 피연산자 값** 자체를 반환한다.

```python
print(9 and 10)    # 10 (9는 참이므로 뒤 값 반환)
print(0 and 10)    # 0  (앞이 거짓이라 뒤는 안 봄)
print(0 or 10)     # 10 (앞이 거짓이라 뒤 평가)
print('a' or 'b')  # 'a' (앞이 참이라 뒤 안 봄)
```

> 💡 자주 쓰는 패턴: `value = user_input or '기본값'` — 사용자 입력이 비었으면 기본값.

## 5. 비트 연산자

정수를 2진법으로 변환해 비트 단위로 연산한다. AI에서는 마스킹, 권한 비트 등에 쓰인다.

| 연산자 | 의미 |
|--------|------|
| `&` | AND (둘 다 1이면 1) |
| `|` | OR (하나라도 1이면 1) |
| `~` | NOT (비트 반전) |
| `<<` | 왼쪽 시프트 (× 2) |
| `>>` | 오른쪽 시프트 (÷ 2) |

```python
print(bin(9), bin(10))       # 0b1001 0b1010
print(bin(9 & 10), 9 & 10)   # 0b1000 8
print(bin(9 | 10), 9 | 10)   # 0b1011 11
print(bin(~9),    ~9)         # -(9+1) = -10

print(4 << 1, 8 << 1)   # 8, 16   (× 2)
print(18 >> 1, 9 >> 1)  # 9, 4    (÷ 2)
```

> ⚠️ `and` 와 `&` 는 다르다.
> - `and` : 논리 연산 (`bool` 기반, 단락 평가)
> - `&` : 비트 연산 (정수의 각 비트별 AND)

```python
print(9 and 10)   # 10 (논리)
print(9 & 10)     # 8  (비트: 1001 & 1010 = 1000)
```

## ✏️ 연습문제

**Q1.** `score = 85` 일 때 `score >= 60 and score < 90` 의 결과는?

**Q2.** 근의 공식으로 $x^2 - 4x + 1 = 0$ 의 두 근을 구해보자.

```python
# Q2 풀이
a, b, c = 1, -4, 1
discriminant = b**2 - 4*a*c
x1 = (-b + discriminant ** 0.5) / (2 * a)
x2 = (-b - discriminant ** 0.5) / (2 * a)
print(x1, x2)
```
