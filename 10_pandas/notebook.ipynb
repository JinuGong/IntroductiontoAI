## 2. DataFrame 만들기

```python
import pandas as pd

# 딕셔너리로 만들기 — 가장 흔한 방법
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Carol', 'Dave'],
    'age':  [30, 25, 28, 35],
    'city': ['Seoul', 'Busan', 'Incheon', 'Daegu'],
    'score': [85, 92, 78, 88],
})
print(df)
```

### 2.1 첫/끝 행 보기, 정보 확인

```python
print(df.head(2))      # 상위 2행
print(df.tail(2))      # 하위 2행
print('-' * 40)
print(df.shape)        # (행, 열)
print(df.columns)      # 열 이름
print(df.dtypes)       # 각 열의 자료형
```

---
## 3. CSV 읽기 / 쓰기

```python
# 쓰기
df.to_csv('people.csv', index=False)        # index=False: 행 번호 저장 안 함

# 읽기
df2 = pd.read_csv('people.csv')
print(df2)

# 인덱스 열을 지정해서 읽기 (예: 'name' 을 인덱스로)
df3 = pd.read_csv('people.csv', index_col='name')
print(df3)
```

> 💡 한글 CSV가 깨질 때: `pd.read_csv(path, encoding='cp949')`

## 4. 슬라이싱 — `loc` 와 `iloc`

| 방식 | 의미 | 예 |
|------|------|-----|
| `df['col']` | 열 선택 → Series |
| `df[['c1','c2']]` | 여러 열 → DataFrame |
| `df[2:5]` | **행** 슬라이스 (정수 위치) |
| `df.loc[label]` | **라벨** 기반 행 선택 |
| `df.iloc[i]` | **위치(정수)** 기반 행 선택 |
| `df.loc[r, c]` | 라벨 행, 라벨 열 |

```python
# 열 선택
print(df['name'])
print(df[['name', 'score']])

# 행 선택 — 정수 슬라이스
print(df[1:3])
```

```python
# 인덱스를 'name' 으로 둔 버전으로 loc 시연
df_idx = df.set_index('name')

# 라벨로 행 한 줄
print(df_idx.loc['Alice'])
print('-' * 30)

# 라벨 + 열
print(df_idx.loc['Alice', 'city'])
print(df_idx.loc[['Alice','Bob'], ['city','score']])
```

```python
# iloc: 정수 위치
print(df.iloc[0])         # 첫 행
print(df.iloc[0:2, 1:3])  # 0~1행, 1~2열
```

## 5. 새 열 추가 · 조건 필터링

새 열은 단순 대입으로 추가. 다른 열을 이용한 계산이 흔하다.

```python
df['score_norm'] = df['score'] / 100     # 0~1 정규화
df['adult'] = df['age'] >= 20            # bool 열
print(df)
```

### 5.1 조건 필터링 — 부울 인덱싱

```python
# 점수가 85 이상인 행
print(df[df['score'] >= 85])

# 여러 조건 — &(and), |(or). 각 조건은 괄호로 감쌀 것
print(df[(df['score'] >= 85) & (df['age'] < 30)])
```

## 6. 통계와 정렬

```python
print(df.describe())      # 수치형 열에 대한 요약 통계
print('-' * 40)
print(df['age'].mean(), df['age'].max(), df['age'].min())
print(df['score'].sum())
```

```python
# 점수 내림차순 정렬
print(df.sort_values('score', ascending=False))

# 두 열로 정렬 (도시 오름차순, 같은 도시면 나이 내림차순)
print(df.sort_values(['city', 'age'], ascending=[True, False]))
```

## 7. 결측값 (NaN) 처리

실제 데이터에는 **빈 칸(NaN)** 이 자주 있다. 그대로 두면 평균 계산 등이 망가진다.

```python
import numpy as np
import pandas as pd

weather = pd.DataFrame({
    '일자':      ['2025-01-01', '2025-01-02', '2025-01-03', '2025-01-04'],
    '평균기온':  [-3.2, -1.5, np.nan, 0.8],
    '평균풍속':  [2.1, np.nan, 3.4, 2.7],
})
print(weather)
print('-' * 40)

# 결측 여부
print(weather.isna())
print('-' * 40)
print('결측 합계:\n', weather.isna().sum())
print('결측 있는 행:\n', weather[weather.isna().any(axis=1)])
```

### 7.1 `fillna` — 결측을 다른 값으로 채우기

```python
# 0 으로 채우기
print(weather.fillna(0))
print('-' * 40)

# 평균으로 채우기 — 열마다 다르게
mean_vals = weather[['평균기온', '평균풍속']].mean()
print(weather.fillna(mean_vals))
```

### 7.2 `dropna` — 결측 있는 행 삭제

```python
print(weather.dropna())                # 결측 하나라도 있으면 그 행 삭제
print('-' * 40)
print(weather.dropna(subset=['평균기온']))  # 특정 열에 결측 있는 행만 삭제
```

## 8. 미니 프로젝트 — 축구 통계 분석

경기당 평균 골 수가 0.75골 이상인 선수를 찾아보자.

```python
import pandas as pd

players = pd.DataFrame({
    'name':    ['Messi', 'Ronaldo', 'Son', 'Mbappe', 'Lewa', 'Kane'],
    'matches': [40, 38, 35, 42, 36, 39],
    'goals':   [38, 32, 23, 36, 33, 24],
})

players['goals_per_game'] = players['goals'] / players['matches']
hot = players[players['goals_per_game'] >= 0.75]
print(hot.sort_values('goals_per_game', ascending=False))
```
