## 1. `for` 와 `range`

`for` 는 **반복 가능한 객체(iterable)** 의 원소를 하나씩 꺼내 변수에 담는다.
`range(n)` 은 0부터 n-1까지의 정수 시퀀스를 만든다.

```python
for t in range(5):
    print(t)
# 출력: 0 1 2 3 4
```

### 1.1 `range` 의 세 가지 형태

| 호출 | 의미 | 예시 |
|------|------|------|
| `range(stop)` | 0부터 stop 미만 | `range(5)` → 0,1,2,3,4 |
| `range(start, stop)` | start부터 stop 미만 | `range(2, 5)` → 2,3,4 |
| `range(start, stop, step)` | step 간격 | `range(5, 0, -1)` → 5,4,3,2,1 |

```python
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(5, 0, -1)))
print(list(range(0, 10, 2)))   # 짝수
```

> 💡 `range(5, 0, 1)` 처럼 **step 방향이 안 맞으면** 빈 시퀀스가 된다.

### 1.2 리스트도 직접 순회

```python
for i in [1, 2, 3, 4, 5]:
    print(i ** 2)
```

## 2. `for` 응용: 1~100의 홀수 합

```python
# 방법 1: 1부터 100까지 돌면서 홀수만
total = 0
for i in range(1, 101):
    if i % 2 == 1:
        total += i
print(total)

# 방법 2: 홀수만 직접 만들기 (더 간결)
total = 0
for i in range(1, 101, 2):
    total += i
print(total)

# 방법 3: 파이썬다움
print(sum(range(1, 101, 2)))
```

## 3. `while` 반복문

**조건이 참인 동안** 반복. 종료 조건을 직접 갱신해야 무한 루프를 피할 수 있다.

```python
t = 0
while t < 5:
    print(t)
    t += 1     # ⚠️ 이걸 깜빡하면 무한 루프
```

### 3.1 1~100 홀수 합 (while 버전)

```python
total = 0
i = 1
while i <= 100:
    total += i
    i += 2     # 홀수만 더하기 위해 2씩 증가
print(total)
```

### 3.2 비밀번호 확인

```python
# 정답을 입력할 때까지 반복
password = ''
while password != 'pythonisfun':
    password = 'pythonisfun'   # = input("비밀번호: ")
    if password != 'pythonisfun':
        print('틀렸습니다. 다시!')
print('로그인 성공')
```

## 4. `break` 와 `continue`

| 키워드 | 동작 |
|--------|------|
| `break` | 반복문을 **즉시 종료** |
| `continue` | 이번 반복은 건너뛰고 **다음 반복으로** |

```python
# break: 첫 짝수를 찾으면 멈춤
for n in [3, 7, 2, 9, 4]:
    if n % 2 == 0:
        print('첫 짝수:', n)
        break
```

```python
# continue: 모음만 건너뛰고 출력
word = 'pythonbasic'
for ch in word:
    if ch in 'aeiou':
        continue
    print(ch, end=' ')
```

### 4.1 숫자 맞히기 (재시도 + 횟수 제한)

```python
import random

answer = random.randint(1, 100)
max_try = 10
tries = 0

while tries < max_try:
    guess = random.randint(1, 100)   # = int(input("1~100: "))
    tries += 1
    if guess == answer:
        print(f'정답! {tries}번 만에 맞혔습니다.')
        break
    elif guess < answer:
        print(f'[{tries}/{max_try}] 더 큰 수')
    else:
        print(f'[{tries}/{max_try}] 더 작은 수')
else:
    # while-else: break 없이 끝나면 실행
    print(f'기회 초과. 정답은 {answer}')
```

> 💡 `while-else` / `for-else` : 반복이 **`break` 없이** 정상 종료될 때 실행되는 블록.

## 5. 미니 프로젝트

### 5.1 Hangman

프로그램이 단어를 고르고, 플레이어는 글자를 추측해 단어를 맞춘다.

```python
import random

words = ['python', 'data', 'science', 'machine', 'learning']
secret = random.choice(words)
revealed = ['_'] * len(secret)
chances = 7
guessed = set()

# 비대화형 데모: 실제로는 input() 으로 받음
fake_guesses = list('aeionrstpyhc')   # 데모용 추측 순서

for ch in fake_guesses:
    if chances == 0 or '_' not in revealed:
        break
    if ch in guessed:
        continue
    guessed.add(ch)

    if ch in secret:
        for i, c in enumerate(secret):
            if c == ch:
                revealed[i] = ch
        print(f'✅ {ch} 있음 →', ' '.join(revealed))
    else:
        chances -= 1
        print(f'❌ {ch} 없음 (남은 기회 {chances}) →', ' '.join(revealed))

print('단어:', secret, '/ 결과:', '성공' if '_' not in revealed else '실패')
```

### 5.2 베스킨라빈스 31 (사람 vs 컴퓨터)

1부터 시작해 번갈아 1~3개씩 숫자를 부르고, 31을 부르는 쪽이 진다.

```python
import random

current = 0
turn = 'me'

while current < 31:
    if turn == 'me':
        # 보통은 input() 으로 받지만 데모에선 작은 값
        n = min(31 - current, random.randint(1, 3))
        print(f'나: {n}개')
    else:
        n = min(31 - current, random.randint(1, 3))
        print(f'컴퓨터: {n}개')

    for _ in range(n):
        current += 1
        print(' ', current, end='')
    print()

    if current >= 31:
        loser = '나' if turn == 'me' else '컴퓨터'
        print(f'💥 {loser}가 31을 불렀습니다. 패배!')
        break
    turn = 'cpu' if turn == 'me' else 'me'
```
