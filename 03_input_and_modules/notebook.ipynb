---
## 1. `input()` 함수

키보드 입력을 받는다. 괄호 안에 안내 메시지를 넣을 수 있다.

```python
name = input("이름을 입력하세요: ")
print(f'안녕하세요, {name}님!')
```

---
## 2. 입력값의 자료형 변환

**`input()` 의 반환값은 항상 `str`** 이다. 숫자로 쓰려면 변환이 필요하다.

```python
age_str = input("나이: ")
print(type(age_str))   # <class 'str'>

age = int(age_str)     # 정수로 변환
print(age + 1, '살이 되었네요')
```

> 💡 한 줄로: `age = int(input("나이: "))`

---
## 3. 여러 값 동시 입력

공백으로 구분된 여러 값을 한 번에 받으려면 **`split()`** 으로 잘라낸다.
전부 숫자라면 **`map()`** 으로 일괄 변환.

```python
# 문자열로 받는 경우
line = input("이름 나이 전화번호: ")
name, age, phone = line.split()
print(name, age, phone)

# 숫자 여러 개 — 한 줄에 1 2 3 입력 시
# a, b, c = map(int, input("세 정수: ").split())
# print(a + b + c)
```

### 3.1 연습: 초 → 시·분·초

```python
# 사용자가 시간을 초로 입력하면 시/분/초로 변환
seconds = 3725  # = int(input("초: "))
h = seconds // 3600
m = (seconds % 3600) // 60
s = seconds % 60
print(f'{h}시간 {m}분 {s}초')
```

### 3.2 연습: 원리금 계산

원금 `a`, 이자율 `r`, `n` 년 후의 원리금 합계는 $a \cdot (1+r)^n$ .

```python
a = 1_000_000   # 원금
r = 0.05        # 5%
n = 10
total = a * (1 + r) ** n
print(f'{n}년 후 원리금 합계: {total:,.0f}원')
```

## 4. `random` 모듈

외부 기능을 쓰려면 먼저 **`import`** 한다.

```python
import random

print(random.randint(1, 45))     # 1~45 사이 정수 (양 끝 포함)
print(random.randrange(7))       # 0~6 사이 정수 (7 미만)

lst = [10, 20, 30, 40, 50]
print(random.choice(lst))        # 리스트에서 하나 무작위 선택
random.shuffle(lst)              # 리스트를 직접 섞음
print(lst)

print(random.random())           # 0.0 ~ 1.0 사이 실수
print(random.uniform(1.0, 2.0))  # 1.0 ~ 2.0 사이 실수
```

## 5. `math` 모듈

```python
import math

print(math.pi)              # 원주율
print(math.e)               # 자연상수
print(math.sqrt(2))         # √2
print(math.log(math.e))     # ln(e) = 1
print(math.sin(math.pi / 2))   # sin(π/2) = 1
print(math.floor(3.7), math.ceil(3.2))   # 내림 3, 올림 4
```

## ✏️ 연습문제

**Q1.** 삼각형의 세 변의 길이를 입력받아 넓이를 구하라 (헤론의 공식: $s = (a+b+c)/2$, $S = \sqrt{s(s-a)(s-b)(s-c)}$ ).

```python
import math
a, b, c = 3, 4, 5    # = map(int, input().split())
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print(f'넓이: {area}')
```
