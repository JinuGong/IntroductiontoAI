## 1. 리스트 만들기 · 원소 추가

대괄호 `[]` 안에 쉼표로 원소를 나열한다. 빈 리스트는 `[]` 또는 `list()`.

```python
bts = ['V', 'Jungkook', 'RM', 'Jin', 'Suga', 'J-Hope', 'Jimin']
print(bts)

empty1 = []
empty2 = list()
print(empty1, empty2)

# range / 문자열로부터 리스트 만들기
print(list(range(5)))
print(list('python'))
```

### 1.1 `append` 와 `+`

```python
bts = []
bts.append('RM')                                # 한 개 추가
bts = bts + ['Jungkook', 'Suga', 'J-Hope']      # 리스트끼리 연결
bts.extend(['Jimin', 'V', 'Jin'])               # extend 도 연결
print(bts)
```

> 💡 `append(x)` 는 x **하나**를 추가, `extend([a,b])` 는 **원소들**을 풀어서 추가.

## 2. 리스트 연산

```python
a = ['V', 'J-Hope'] + ['RM', 'Jin']
print(a)                # 연결

b = [0, 1, 2] * 3
print(b)                # 반복

print('V' in a)         # 포함 확인
print('Jinu' not in a)
```

## 3. 인덱싱과 슬라이싱

인덱스는 **0부터** 시작. 음수 인덱스는 **뒤에서부터** 센다.

```python
letters = ['A', 'B', 'C', 'D', 'E', 'F']
print(letters[0])     # 'A' — 첫 원소
print(letters[-1])    # 'F' — 마지막 원소
print(letters[-2])    # 'E' — 끝에서 두 번째
```

### 3.1 슬라이싱 `lst[start:stop:step]`

| 표기 | 의미 |
|------|------|
| `lst[2:5]` | 인덱스 2~4 (5 미만) |
| `lst[2:]`  | 인덱스 2부터 끝까지 |
| `lst[:3]`  | 처음부터 인덱스 2까지 |
| `lst[::2]` | 처음부터 끝까지 2칸씩 |
| `lst[::-1]`| 역순 |

```python
letters = ['A', 'B', 'C', 'D', 'E', 'F']
print(letters[2:5])    # ['C', 'D', 'E']
print(letters[2:])     # ['C', 'D', 'E', 'F']
print(letters[::2])    # ['A', 'C', 'E']
print(letters[::-1])   # ['F', 'E', 'D', 'C', 'B', 'A']
```

### 3.2 값 변경 · 삭제

```python
slist = ['Kim', 178.9, 'Park', 173.5, 'Lee', 176.1]

slist[3] = 180.0           # 인덱스 3 변경
print(slist)

del slist[0]               # 인덱스 0 삭제
print(slist)

slist.remove('Lee')        # 값으로 첫 번째 매칭 삭제
print(slist)
```

### 3.3 연습: 문자열에서 특정 단어 제거

```python
text = 'Python is love'
words = text.split()
print(words)               # ['Python', 'is', 'love']

words[-1] = 'fun'
print(' '.join(words))     # 'Python is fun'
```

## 4. 정렬과 자주 쓰는 함수

| 함수/메서드 | 의미 |
|-------------|------|
| `len(lst)` | 원소 개수 |
| `min(lst)`, `max(lst)`, `sum(lst)` | 최솟값/최댓값/합 |
| `lst.sort()` | **제자리** 정렬 (반환 없음) |
| `sorted(lst)` | 정렬된 **새 리스트** 반환 |
| `lst.reverse()` | 제자리 뒤집기 |

```python
n_list = [100, 200, 300, 400, 450]
print(len(n_list), min(n_list), max(n_list), sum(n_list))

a = [5, 3, 2, 1, 4]
print(sorted(a))           # [1,2,3,4,5] — 원본은 그대로
print(a)                   # [5,3,2,1,4]

a.sort()                   # 원본을 직접 변경
print(a)

a.sort(reverse=True)
print(a)
```

## 5. 객체와 참조 — `=` 는 복사가 아니다

리스트를 다른 변수에 대입하면 **같은 객체**를 가리키게 된다. 한쪽을 바꾸면 다른 쪽도 바뀐다.

```python
a = [1, 2, 4]
b = a                # 같은 리스트를 가리킴
b[0] = 99
print(a, b)          # [99, 2, 4] [99, 2, 4]
print(id(a), id(b))  # 동일 — 같은 객체
```

### 5.1 진짜 복사하려면

```python
a = [1, 2, 4]
b = a.copy()         # 또는 list(a), a[:]
b[0] = 99
print(a, b)          # [1, 2, 4] [99, 2, 4] — 독립적
```

## 6. 리스트 컴프리헨션

한 줄로 리스트를 생성하는 표현식. **`[표현식 for 변수 in 시퀀스 if 조건]`**

```python
squares = [x ** 2 for x in range(1, 6)]
print(squares)       # [1, 4, 9, 16, 25]

# 조건 필터링
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
print(even_squares)  # [0, 4, 16, 36, 64]

# 짝수만 추출
nums = [1, 2, 3, 4, 5, 6]
print([n for n in nums if n % 2 == 0])
```

### 6.1 모든 'banana' 제거하기

`remove` 는 **첫 번째**만 지운다. 전부 지우려면 컴프리헨션이 깔끔.

```python
fruits = ['apple', 'banana', 'cherry', 'banana']
no_banana = [f for f in fruits if f != 'banana']
print(no_banana)
```

### 6.2 연습: 2부터 n까지의 소수 리스트

```python
def primes(n: int) -> list[int]:
    """에라토스테네스의 체로 2~n의 소수 반환."""
    sieve = [True] * (n + 1)
    sieve[0] = sieve[1] = False
    for i in range(2, int(n ** 0.5) + 1):
        if sieve[i]:
            for j in range(i * i, n + 1, i):
                sieve[j] = False
    return [i for i, ok in enumerate(sieve) if ok]

print(primes(50))
```

## 7. 튜플 — 변경 불가능한 시퀀스

튜플은 소괄호 `()` 로 만든다. **원소 변경이 불가능** 한 점만 빼면 리스트와 비슷하다.

```python
colors = ('r', 'g', 'b')
print(colors[0])

# colors[0] = 'k'   # ❌ TypeError — 튜플은 변경 불가

# 새 튜플을 만들어 합칠 수는 있음
new_color = colors + ('c', 'k')
print(new_color)

# 단일 원소 튜플은 쉼표 필수: (1,)
single = (1,)
not_tuple = (1)
print(type(single), type(not_tuple))
```

> 💡 **언제 튜플?** 좌표 `(x, y)`, 함수에서 여러 값 반환, 딕셔너리 key 등 **바뀌면 안 되는 묶음** 표현.
