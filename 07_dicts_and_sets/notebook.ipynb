## 1. 딕셔너리 기본

중괄호 `{}` 안에 `키: 값` 쌍을 쉼표로 나열. 키는 보통 문자열이나 숫자.

```python
# 빈 딕셔너리
empty = {}
empty2 = dict()

# 초기값 있음
person = {"name": "Alice", "age": 30, "city": "Seoul"}
print(person)
print(person["name"])     # 키로 값 조회
```

---
## 2. CRUD — 추가, 수정, 삭제, 조회

```python
person = {"name": "Alice", "age": 30, "city": "Seoul"}

# 추가
person["job"] = "Engineer"

# 수정 (같은 키에 새 값)
person["age"] = 31

# 삭제
del person["city"]

# 조회 — 키가 없으면 KeyError
print(person["name"])

# 안전 조회: 없으면 None / 기본값
print(person.get("phone"))
print(person.get("phone", "없음"))

# 키 존재 확인
print("name" in person)
print("phone" in person)
```

---
## 3. 자주 쓰는 메서드

| 메서드 | 설명 |
|--------|------|
| `d.keys()` | 키만 |
| `d.values()` | 값만 |
| `d.items()` | `(키, 값)` 쌍 |
| `d.get(key, default)` | 안전 조회 |
| `d.pop(key)` | 키-값 제거하고 값 반환 |
| `d.popitem()` | 마지막 쌍 제거하고 반환 |
| `d.update(other)` | 다른 딕셔너리 병합 |

```python
person = {"name": "Alice", "age": 30, "city": "Seoul"}
print(list(person.keys()))
print(list(person.values()))
print(list(person.items()))

age = person.pop("age")
print(age, person)

key, value = person.popitem()
print(key, value, person)
```

---
## 4. 딕셔너리 순회

키, 값, 둘 다 어떻게 순회할 수 있는지 알아두면 좋다.

```python
scores = {"Alice": 85, "Bob": 92, "Charlie": 78}

# (a) 키만
for name in scores:
    print(name, scores[name])

# (b) 키-값 쌍 동시에 (가장 흔한 패턴)
for name, score in scores.items():
    print(f'{name}: {score}')

# (c) 값만 필요할 때
total = sum(scores.values())
print('총점:', total)
```

## 5. 집합 (`set`)

중복 없는 모음. **순서가 없고** 인덱싱이 안 된다. 멤버십 검사가 빠르다.

```python
# 리스트의 중복 제거
int_list = [1, 2, 3, 4, 4, 5, 5]
unique = set(int_list)
print(unique)               # {1, 2, 3, 4, 5}
print(list(unique))         # 다시 리스트로

# 추가/삭제
s = {1, 2, 3}
s.add(4)
s.remove(1)
print(s)
```

### 5.1 합집합 · 교집합 · 차집합

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)      # 합집합 {1,2,3,4,5,6}
print(a & b)      # 교집합 {3,4}
print(a - b)      # 차집합 {1,2}
print(a ^ b)      # 대칭차 {1,2,5,6}
```

> 💡 멤버십 검사 `x in s` 는 리스트보다 훨씬 빠르다 (해시 기반).

## 6. 미니 프로젝트

### 6.1 단어 빈도 분석

텍스트에서 각 단어가 몇 번 나오는지 센다.

```python
text = '''The quick brown fox jumps over the lazy dog.
The dog barks. The fox runs away.'''

# 1) 소문자 변환 + 구두점 제거
import string
clean = text.lower().translate(str.maketrans('', '', string.punctuation))
words = clean.split()

# 2) 빈도 카운트
freq = {}
for w in words:
    freq[w] = freq.get(w, 0) + 1

# 3) 빈도 높은 순으로 정렬
top = sorted(freq.items(), key=lambda kv: -kv[1])
for word, n in top[:5]:
    print(f'{word:10s} {n}')
```

> 💡 `collections.Counter` 를 쓰면 더 짧다 — `Counter(words).most_common(5)`

### 6.2 편의점 재고 관리

상품명을 키, 수량을 값으로 저장해 입출고를 관리한다.

```python
inventory = {}

def stock_in(item: str, qty: int):
    inventory[item] = inventory.get(item, 0) + qty

def stock_out(item: str, qty: int):
    if inventory.get(item, 0) < qty:
        print(f'❌ {item} 재고 부족 (현재 {inventory.get(item, 0)})')
        return
    inventory[item] -= qty
    if inventory[item] == 0:
        del inventory[item]

def show():
    if not inventory:
        print('재고 없음')
        return
    for item, qty in sorted(inventory.items()):
        print(f'  {item:8s} : {qty}개')

stock_in('콜라', 10)
stock_in('사이다', 5)
stock_in('콜라', 3)
stock_out('콜라', 4)
stock_out('맥주', 1)   # 재고 없음
show()
```
