# 🔀 조건문

**주제**: `if`, `elif`, `else` 로 조건에 따라 다른 코드를 실행하기

| 섹션 | 내용 |
|------|------|
| 1. `if` 기본 형식 | 조건이 참이면 실행 |
| 2. `if-else` | 조건이 거짓일 때 대안 |
| 3. `if-elif-else` | 여러 조건 분기 |
| 4. 들여쓰기 규칙 | 블록을 구분하는 핵심 |
| 5. 미니 프로젝트 | 계산기 · BMI · 복권 · 숫자 맞히기 |


---
## 1. `if` 기본 형식

```python
if 조건:
    실행할_코드
```

- **조건 뒤의 콜론 `:`** 잊지 말 것.
- 조건이 참(`True`)이면 들여쓰여진 코드가 실행된다.

````python
score = 85
if score >= 90:
    print("A")
print("프로그램 종료")   # 들여쓰기 밖이라 항상 실행
````

> ⚠️ **들여쓰기**가 곧 블록 구분이다. 들여쓰기 칸 수는 자유지만 **같은 블록 안에서는 일관성** 있게.
> 권장: 스페이스 4칸 (PEP8), Colab은 2칸 기본.

---
## 2. `if-else`

조건이 거짓일 때 실행할 코드를 `else` 블록에 둔다.

````python
score = 75
if score >= 90:
    print("A")
else:
    print("A 아님")
````

---
## 3. `if-elif-else`

여러 조건을 차례로 검사하려면 `elif` (else if). 위에서부터 처음 참이 되는 조건만 실행된다.

````python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("F")
````

### 3.1 `elif` 없이 `if` 만으로 쓰면?

````python
# ❌ 비효율적: 모든 if를 다 검사하고, 두 조건이 동시에 참이면 둘 다 실행됨
score = 85
if score >= 90: print("A")
if score >= 80: print("B")   # 85 ≥ 80 도 참이라 출력됨
if score >= 70: print("C")
````

👉 학점은 한 등급만 나와야 하므로 `elif` 가 맞다.

## 4. 미니 프로젝트

### 4.1 사칙연산 계산기

````python
a, b = 7, 3
op = '+'   # = input("연산자(+-*/): ")

if op == '+':
    print(a + b)
elif op == '-':
    print(a - b)
elif op == '*':
    print(a * b)
elif op == '/':
    if b == 0:
        print("0으로 나눌 수 없습니다")
    else:
        print(a / b)
else:
    print("지원하지 않는 연산자")
````

### 4.2 BMI 계산기

$BMI = \dfrac{\text{체중(kg)}}{\text{키(m)}^2}$

````python
height_cm = 170
weight_kg = 65
h_m = height_cm / 100
bmi = weight_kg / (h_m ** 2)

if bmi < 18.5:
    grade = '저체중'
elif bmi < 25:
    grade = '정상'
elif bmi < 30:
    grade = '과체중'
else:
    grade = '비만'

print(f'BMI: {bmi:.2f} ({grade})')
````

### 4.3 복권 (3개 숫자 일치 여부)

````python
import random

my = (3, 17, 25)
win = (random.randint(1, 30), random.randint(1, 30), random.randint(1, 30))

print('내 번호:', my)
print('당첨 번호:', win)

if my == win:
    print('🎉 1등!')
elif set(my) == set(win):
    print('🥈 순서만 다름')
else:
    matched = len(set(my) & set(win))
    print(f'일치 개수: {matched}')
````

### 4.4 숫자 맞히기 (한 번 시도)

````python
import random

answer = random.randint(1, 100)
guess  = 50   # = int(input("1~100 사이 숫자: "))

if guess == answer:
    print('정답!')
elif guess < answer:
    print(f'더 큰 수입니다 (정답은 {answer})')
else:
    print(f'더 작은 수입니다 (정답은 {answer})')
````
